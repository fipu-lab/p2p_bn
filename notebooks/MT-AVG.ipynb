### Reddit -> Few-NERD

```python
from environ import set_visible_devices
set_visible_devices('GPU:0')

from p2p.train import do_train
from p2p.agents import BertAgent
from data.ner import clients_data as ner_clients_data
from models.ner import model as ner_model
from data.reddit import bert_clients_data
from models.nwp import bert_model as nwp_model

seed = 123

do_train(
    agent_pars=[
        {
            "agent_class": BertAgent,
            "multi_task_optimization": True,
            "multi_output_averaging": True,
            "multi_task_differentiation": "model",
            "use_tf_function": False,
        },
        {
            "agent_class": BertAgent,
            "multi_task_optimization": True,
            "multi_output_averaging": True,
            "multi_task_differentiation": "model",
            "use_tf_function": False,
        },
    ],
    agent_data_pars=[
        {
            "agents_data": bert_clients_data,
            "num_clients": 20,
            "batch_size": 50,
            "seq_len": 128,
            "seed": seed,
        },
        {
            "agents_data": ner_clients_data,
            "num_clients": 20,
            "batch_size": 50,
            "dataset": "few_nerd",
            "dataset_fraction": 0.2,
        },
    ],
    model_pars=[
        {
            "model_mod": nwp_model,
            "bert_config": "uncased_L-2_H-128_A-2",
            "seq_len": 128,
            "lr": 5e-4,
            "default_weights": "global",
        },
        {
            "model_mod": ner_model,
            "bert_config": "uncased_L-2_H-128_A-2",
            "processor_name": "few",
            "lr": 5e-4,
            "default_weights": "global",
        },
    ],
    graph_pars={'graph_type': 'sparse_clusters', 'num_neighbors': 2, 'cluster_conns': 1, 'cluster_directed': True, 'directed': True, 'time_varying': -1},
    sim_pars={
        "epochs": 300,
        "seed": seed,
        "print_args": {
            "print_metrics": [
                "test_model-macro_avg_f1_score",
                "test_model-sparse_categorical_accuracy",
            ],
            "group_by_dataset": False,
        },
    },
)
```
